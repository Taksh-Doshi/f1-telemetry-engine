```python
import os
from pathlib import Path

import fastf1
from dotenv import load_dotenv

project_root = Path.cwd().resolve()
while not (project_root / ".env.example").is_file() and project_root.parent != project_root:
    project_root = project_root.parent

load_dotenv(project_root / ".env")
cache_dir = Path(os.getenv("FASTF1_CACHE_DIR", "cache"))
if not cache_dir.is_absolute():
    cache_dir = project_root / cache_dir

fastf1.Cache.enable_cache(str(cache_dir))
```

```python
session = fastf1.get_session(2025, "Monza", "R")
session.load()
```

```python
session.laps
```

```python
session.laps.pick_driver("VER")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['figure.figsize'] = [10, 6]
laps = session.laps.pick_driver("VER")
speed_columns =  {
    "SpeedI1": "Intermediate 1",
    "SpeedI2": "Intermediate 2",
    "SpeedFL": "Finish Line",
    "SpeedST": "Speed Trap",
}
for column in speed_columns:
    plt.plot(laps["LapNumber"], laps[column], marker="o", label=speed_columns[column])

plt.xlabel("Lap Number")
plt.ylabel("Speed (km/h)")
plt.title("Verstappen speed by lap")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
